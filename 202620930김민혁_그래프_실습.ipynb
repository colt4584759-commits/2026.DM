```python
# 예시 그래프: 친구관계
graph = {
    '김광수': ['파이리', '꼬부기', '이상해씨', '김민석', '김민혁'],
    '이윤건': ['박서준', '왕현식', '김민혁', '곽정훈', '박형준'],
    '김민석': ['최우진', '안우현', '이재용', '심문석', '홍준서'],
    '김민혁': ['곽정훈', '안우현', '김민준', '최우진'],
    '곽정훈': ['최우진', '안우현', '배정우',],
    '파이리': [],
    '꼬부기': [],
    '이상해씨': [],
    '박서준': [],
    '왕현식': [],
    '박형준': [],
    '최우진': [],
    '안우현': [],
    '김민준': [],
    '이재용': [],
    '심문석': [],
    '홍준서': [],
    '배정우': []
}

from collections import deque

def bfs(graph, start):
  visited = set()
  queue = deque([start])
  visited.add(start)

  while queue:
    v = queue.popleft()
    print(v, end=' ')
    for neighbor in graph[v]:
      if neighbor not in visited:
        visited.add(neighbor)
        queue.append(neighbor)

def recommend_friend(graph, start):
  visited = set()
  queue = deque([(start,0)]) # (노드, 거리)

  print(f"\n감염자 {start}의 친구 (1단계):")

  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 1:
        print("-", current)
      if depth < 1:
        for neighbor in graph[current]:
          queue.append((neighbor, depth + 1))

  visited = set()
  queue = deque([(start,0)]) # (노드, 거리)

  print(f"\n감염자 {start}의 친구 (2단계):")

  while queue:
    current, depth = queue.popleft()
    if current not in visited:
      visited.add(current)
      if depth == 2:
        print("-", current)
      if depth < 2:
        for neighbor in graph[current]:
          queue.append((neighbor, depth + 1))

#print("BFS 탐색순서: ")
#bfs(graph, '김광수')

recommend_friend(graph, '김민혁')
```
